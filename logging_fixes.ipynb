```python
import logging
import os

LOG_FORMAT = "%(asctime)s - %(name)s - %(levelname)s - %(message)s"
os.makedirs("logs", exist_ok=True)

logging.basicConfig(level=logging.DEBUG, format=LOG_FORMAT, force=True)

bad_logger = logging.getLogger("bad_module")
bad_logger.addHandler(logging.StreamHandler())
bad_logger.error("this message shows up twice")
```

```python
# basicConfig was already called above, so this second call does nothing
logging.basicConfig(filename="logs/ignored.log", level=logging.DEBUG)
print("ignored.log was created:", os.path.exists("logs/ignored.log"))
```

```python
good_logger = logging.getLogger("good_module")
good_logger.setLevel(logging.DEBUG)
good_logger.propagate = False

if not good_logger.handlers:
    handler = logging.StreamHandler()
    handler.setFormatter(logging.Formatter(LOG_FORMAT))
    good_logger.addHandler(handler)

good_logger.error("this message shows up once")
```
